```python
import numpy as np
data=np.array([100,200,300,400,500])
min_value=np.min(data)
max_value=np.max(data)
min_max_normalized=(data-min_value)/(max_value-min_value)
mean=np.mean(data)
std_dev=np.std(data)
z_score_standardized=(data-mean)/std_dev
print("Original data",data)
print("Min-Max Normalized Data:",min_max_normalized)
print("z_Score Normalized Data:",z_score_standardized)
```
